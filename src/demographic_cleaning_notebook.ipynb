# Demographic Data

```python
import pandas as pd
import numpy as np
```

```python
# data years
data_years = [2015, 2016, 2017, 2018, 2019, 2021, 2022, 2023, 2024]

# empty our_data for for loop
our_data = []

# dictionary to change variable names
variable_names = {
    "B01003_001E": "total_population estimate",
    "B01003_001M": "total_population MoE",
    "B19013_001E": "median_HH_income estimate",
    "B19013_001M": "median_HH_income MoE",
    "B25001_001E": "total_housing_units estimate",
    "B25001_001M": "total_housing_units MoE",
    "B25002_003E": "vacant_housing_units estimate",
    "B25002_003E": "vacant_housing_units MoE"
}

# loop to read in all yearly, add year. vacancy_rate_pct, convert to long format
for year in data_years:
    filename = f"acs{year}.csv"
    df = pd.read_csv(filename, header=1)

   # add year to each csv
    df["year"] = year

    # add vacancy_rate_pct
    df["vacancy_rate_pct"] = 100*(df["B25002_003E"]/df["B25002_001E"])

    # convert to long format
    df_long = df.melt(
        id_vars=["NAME", "metropolitan statistical area/micropolitan statistical area", "year"],
        var_name="variable",
        value_name="value"
    )

    our_data.append(df_long)

# combine into one mega set
all_acs = pd.concat(our_data, ignore_index=True)

# change names using dictionary
all_acs["variable"] = all_acs["variable"].replace(variable_names)

# check to see if it looks OK
all_acs.head()

# save as .csv file
all_acs.to_csv("all_acs.csv", index=False)
```
